```python
print("crime_type unique:", df['crime_type'].unique())
```

```python
crime_mapping={
    "asslt":"Assault","assault":"Assault","battery":"Assault","assault & battery":"Assault","homocide":"Homicide",
    "homicide":"Homicide","murder":"Homicide","manslaughter":"Homicide","burglary":"Burglary","burglry":"Burglary",
    "b&e":"Burglary","breaking & entering":"Burglary","robbery":"Robbery","robbry":"Robbery","roberry":"Robbery",
    "armed robbery":"Robbery","theft":"Theft","larceny":"Theft","theft/larceny":"Theft","stealing":"Theft",
    "dv":"Domestic Violence", "dom. violence": "Domestic Violence", "domestc violence":"Domestic Violence",
    "domestic violence": "Domestic Violence","sa":"Sexual Assault", "sex assault": "Sexual Assault", "sexual assault": "Sexual Assault",
    "sexual assualt": "Sexual Assault", "dui":"DUI","duii": "DUI","dwi": "DUI","d.u.i.": "DUI","drunk driving": "DUI",
    "drugs":"Drug Offense","narcotics": "Drug Offense","drug offense": "Drug Offense","drug offence":"Drug Offense",
    "fraud":"Fraud", "fraudulent activity": "Fraud", "online fraud": "Fraud", "scam": "Fraud","deception":"Fraud",
    "arson":"Arson", "arsen":"Arson","fire setting": "Arson", "vandalism": "Vandalism",
    "vandlism":"Vandalism","graffiti":"Vandalism","property damage":"Vandalism","trespass":"Trespassing","trespassing": "Trespassing",
     "tresspassing":"Trespassing", "kidnapping": "Kidnapping","kidnaping":"Kidnapping","abduction":"Kidnapping",
    "cyber crime":"Cybercrime","cybercrime":"Cybercrime","hacking":"Cybercrime",
}

crime_clean_df["crime_type_clean"] = (crime_clean_df["crime_type"]
.astype(str).str.strip().str.lower()
.str.replace(r'\s+', ' ', regex=True)
.map(crime_mapping)
)
```

```python
crime_clean_df.crime_type_clean.unique()
```

```python
print("district unique:", df['district'].unique())
```

```python
district_map={
    "sou":"South",
    "cen":"Central",
    "nor":"North",
    "eas":"East",
    "mid":"Midtown",
    "wes":"West",
}
crime_clean_df["district_clean"]=(
    crime_clean_df["district"]
    .str.strip()
    .str.lower()
    .replace(district_map)
    .str.title()

)
```

```python
print("suspect_gender unique:", df['suspect_gender'].unique())
print("victim_gender unique:", df['victim_gender'].unique())
```

```python
gender_map= {
    "m":"Male", "male":"Male", "f":"Female", "female":"Female", "other":"Other", "unknown":"Unknown",
}
crime_clean_df["suspect_gender_clean"]=(
    crime_clean_df["suspect_gender"]
    .astype(str).str.strip().str.lower().map(gender_map)
)
crime_clean_df["victim_gender_clean"]=(
    crime_clean_df["victim_gender"].astype(str).str.strip().str.lower().map(gender_map)
)
```

```python
#
crime_clean_df["incident_datetime_clean"]=pd.to_datetime(crime_clean_df["incident_datetime"], errors="coerce")
crime_clean_df["year"]=crime_clean_df["incident_datetime_clean"].dt.year
crime_clean_df["month"]=crime_clean_df["incident_datetime_clean"].dt.month
crime_clean_df["day_name"]=crime_clean_df["incident_datetime_clean"].dt.day_name()
crime_clean_df["hour"]=crime_clean_df["incident_datetime_clean"].dt.hour
```
